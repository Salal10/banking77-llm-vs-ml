```python
from dotenv import load_dotenv   
load_dotenv()

from google import genai

client = genai.Client()   

for m in client.models.list():
    print(m.name)
```

```python
MODEL = "gemini-3.5-flash-lite"

response = client.models.generate_content(
    model=MODEL,
    contents="A bank customer wrote: 'my card still hasn't arrived'. In a few words, what is their problem?",
)

print(response.text)
```
