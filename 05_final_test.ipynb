```python
from dotenv import load_dotenv
load_dotenv()

import pandas as pd
from google import genai
from sklearn.metrics import accuracy_score, f1_score
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from banking77_llm_vs_ml.core import load_data, make_schema, build_few_shot_prompt, run_eval

client = genai.Client()
MODEL = "gemini-3.5-flash-lite"
THRESHOLD = 0.3   # chosen on validation, BEFORE looking at test

train_df, val_df, test = load_data()
categories = sorted(train_df["category"].unique()) + ["out_of_scope"]
schema = make_schema(categories)
PROMPT_V2 = build_few_shot_prompt(train_df, per_category=3)   # same prompt as validation

# Final P1: train on train + validation (validation's job is done)
full_train = pd.concat([train_df, val_df])
p1 = make_pipeline(TfidfVectorizer(), LogisticRegression(max_iter=1000, C=10))
p1.fit(full_train["text"], full_train["category"])
```

```python
test = test.copy()
probs = p1.predict_proba(test["text"])
test["p1_pred"] = p1.classes_[probs.argmax(axis=1)]
test["p1_conf"] = probs.max(axis=1)

hard = test[test["p1_conf"] < THRESHOLD]
print(f"{len(hard)} of {len(test)} test messages go to the LLM ({len(hard) / len(test):.1%})")
```

```python
llm = run_eval(client, MODEL, PROMPT_V2, schema, hard["text"].drop_duplicates(),
               "results/final_test_llm_v2_hard.csv")
print("Failed:", llm["pred"].isna().sum())
```

```python
r = test.merge(llm[["text", "pred"]].drop_duplicates("text"), on="text", how="left")
use_llm = (r["p1_conf"] < THRESHOLD) & r["pred"].notna()
r["hybrid_pred"] = r["p1_pred"].where(~use_llm, r["pred"])

hard_rows = r[r["p1_conf"] < THRESHOLD]
final = pd.DataFrame([
    {"system": "P1 alone", "accuracy": accuracy_score(r["category"], r["p1_pred"]),
     "macro_f1": f1_score(r["category"], r["p1_pred"], average="macro"), "llm_calls": 0},
    {"system": f"Hybrid P1 + LLM v2 (threshold {THRESHOLD})",
     "accuracy": accuracy_score(r["category"], r["hybrid_pred"]),
     "macro_f1": f1_score(r["category"], r["hybrid_pred"], average="macro"),
     "llm_calls": int(use_llm.sum())},
]).round(3)
print(final)

print("\nOn the uncertain messages only:")
print("  P1 accuracy: ", round((hard_rows["p1_pred"] == hard_rows["category"]).mean(), 3))
print("  LLM accuracy:", round((hard_rows["pred"] == hard_rows["category"]).mean(), 3))

final.to_csv("results/final_test_summary.csv", index=False)
```
